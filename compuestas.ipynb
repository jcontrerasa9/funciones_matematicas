```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
N = 1000

x = np.linspace(-10, 10, num=N)
```

```python
def g(x):
    return x **2

y = g(x)

plt.plot(x, y)

```

```python
def f(x):
    return np.sin(x)

y = f(x)

plt.plot(x, y)
```

```python
f_o_g = f(g(x))

plt.plot(x, f_o_g)
```
